```python
import pandas as pd

data = {
    "post_id": [1, 2, 3, 4, 5],
    "label": [
        "Reliable",
        "Misinformation",
        "Reliable",
        "Misinformation",
        "Reliable"
    ],
    "likes": [120, 80, 200, 50, 150]
}

df = pd.DataFrame(data)

print(df)
```

```python
label_counts= df["label"].value_counts()
label_percentages= df["label"].value_counts(normalize=True)*100

print("Class count:")
print(label_counts)

print("\nClass Percentage:")
print(label_percentages)
```

```python
print("Missing value")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nDtata set Shaoe:")
print(df.shape)
```

```python
import matplotlib.pyplot as plt
df["likes"].plot(
    kind="hist",
    bins=5,
    title="Distribution of likes",
    edgecolor="Black"
)
plt.xlabel("Number of Likes")
plt.ylabel("Frequency")
plt.show()
```

```python
df.boxplot(column="likes")
plt.title("Box plot of likes")
plt.ylabel("Number of likes")
plt.show()
```

```python
print("Minimum:",df["likes"].min())
print("Maximum:",df["likes"].max())
print("Median:",df["likes"].median())
```

```python
print("Q1: ",df["likes"].quantile(0.25))
print("Q3: ",df["likes"].quantile(0.75))
```

```python
import pandas as pd
engagement_df= pd.DataFrame({
    "views":[100,200,300,400,500,600],
    "likes":[10,20,25,40,45,60]
})
print(engagement_df)
```

```python
import matplotlib.pyplot as plt
engagement_df.plot(
    kind="scatter",
    x="views",
    y="likes",
    title="views vs likes"
)
plt.show()
```

```python
correlation=engagement_df["views"].corr(engagement_df["likes"])
print("correlation:",correlation)
```

```python
engagement_df.corr(numeric_only=True)
```

```python
correlation_matrix=engagement_df.corr(numeric_only=True)
print(correlation_matrix)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
plt.imshow(correlation_matrix,
          cmap="coolwarm",
          vmin=-1,
          vmax=1
          )
plt.colorbar(label="correlation")
plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)
plt.yticks(
    range(len(correlation_matrix.index)),
    correlation_matrix.index
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
print(correlation_matrix.round(2))
```

```python
import pandas as pd
research_df = pd.DataFrame({
    "post_id":[1,2,3,4,5,6,7,8],
    "label":["Reliable","Misinformation","Reliable","Reliable","Misinformation","Reliable","Misinformation","Reliable"],
    "platform":["Facebook","YouTube","Facebook","X","YouTube","Facebook","X","YouTube"]
})
print(research_df)
```

```python
research_df["label"].value_counts()
```

```python
research_df["platform"].value_counts()
```

```python
label_percentage=(
    research_df["label"].value_counts(normalize=True)*100
)
print(label_percentage.round(2))
```

```python
platform_percentage=(
    research_df["platform"].value_counts(normalize=True)*100
)
print(platform_percentage.round(2))
```

```python
import matplotlib.pyplot as plt
research_df["label"].value_counts().plot(
    kind="bar",
    title="Distribution od reserarch labels",
    xlabel="Label",
    ylabel="Number of rosts",
    rot=0
)
plt.tight_layout()
plt.show()
```
